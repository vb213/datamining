```python
product_to_section = dict(
    zip(products_df["product"], products_df["section"])
)

purchase_sections = [
    [product_to_section.get(x, x) for x in sublist]
    for sublist in purchase_transactions
]

purchase_sections

click_sections = [
    [product_to_section.get(x, x) for x in sublist]
    for sublist in click_transactions
]

click_sections
```

```python
click_sections = [list(set(t)) for t in click_sections]
purchase_sections = [list(set(t)) for t in purchase_sections]


te = TransactionEncoder()

clicks_sections_encoded = te.fit(click_sections).transform(
    click_sections,
    sparse=True
)

clicks_sections_encoded = clicks_sections_encoded.toarray()

clicks_sections_encoded = pd.DataFrame(
    clicks_sections_encoded,
    columns=te.columns_
)

purchase_section_encoded = one_hot_encode(purchase_sections)
```
